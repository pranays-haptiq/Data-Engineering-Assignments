75. How to split a text column into two separate columns?

```python
import numpy as np
import pandas as pd

df = pd.DataFrame(["STD, City State",
"33, Kolkata West Bengal",
"44, Chennai Tamil Nadu",
"40, Hyderabad Telengana",
"80, Bangalore Karnataka"], columns=['row'])


df_out = df.row.str.split(',\s+', expand=True)

new_header = df_out.iloc[0]
df_out = df_out[1:]
df_out.columns = new_header
print(df_out)
```

25. How to filter valid emails from a series?

```python
import numpy as np
import pandas as pd
emails = pd.Series(['buying books at amazom.com', 'rameses@egypt.com', 'matt@t.co', 'narendra@modi.com'])
pattern ='[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\\.[A-Za-z]{2,4}'

valid_mask = emails.str.match(pattern)
valid_emails = emails[valid_mask]
print(valid_emails)
```

22. How to get the day of month, week number, day of year and day of week from a series of date strings?

```python
import numpy as np
import pandas as pd

ser = pd.Series(
    [
        "01 Jan 2010",
        "02-02-2011",
        "20120303",
        "2013/04/04",
        "2014-05-05",
        "2015-06-06T12:20",
    ]
)

from dateutil.parser import parse

ser_ts = ser.map(lambda x: parse(x))

# day of month
print("Date: ", ser_ts.dt.day.tolist())

# week number
print("Week number: ", ser_ts.dt.isocalendar().week.tolist())

# day of year
print("Day num of year: ", ser_ts.dt.dayofyear.tolist())

# day of week
print("Day of week: ", ser_ts.dt.day_name().tolist())
```

10. How to keep only top 2 most frequent values as it is and replace everything else as ‘Other’?

```python
import numpy as np
import pandas as pd

np.random.seed(100)
ser = pd.Series(np.random.randint(1, 5, [12]))

print("Top 2 Freq:", ser.value_counts())
ser[~ser.isin(ser.value_counts().index[:2])] = "Other"
print(ser)
```

6. How to get the items of series A not present in series B?

```python
import numpy as np
import pandas as pd

ser1 = pd.Series([1, 2, 3, 4, 5])
ser2 = pd.Series([4, 5, 6, 7, 8])

print(ser1[~ser1.isin(ser2)])
```
